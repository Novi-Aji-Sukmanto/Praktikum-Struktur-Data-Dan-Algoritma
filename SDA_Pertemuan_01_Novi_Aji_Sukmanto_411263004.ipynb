```python
def cari(data, target):
    for i in range(len(data)):
        if data[i] == target:
            return i
    return -1
```

```python
data = [12, 20, 7, 30]
print(cari(data, 7))
print(cari(data, 99))
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
node1 = Node(10)
print(node1.data)
print(node1.next)
```

```python
mahasiswa = [
('22001', 'Alya'),
('22002', 'Bima'),
('22003', 'Citra')
]
def cari_nim(data, nim):
    for item in data:
        if item[0] == nim:
           return item
    return None
print(cari_nim(mahasiswa, '22003'))
```
