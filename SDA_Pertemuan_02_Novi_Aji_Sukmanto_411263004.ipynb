```python
#2.21. Praktik Mandiri 2.2 — Linear Search
# Data berupa 15 bilangan yang tidak terurut
data = [43, 42, 88, 1, 71, 24, 97, 50, 36, 59, 70, 12, 5, 64, 21]

print("Data:", data)

# Fungsi Linear Search dengan trace
def linear_search_trace(data, target):
    jumlah_pemeriksaan = 0

    for i in range(len(data)):
        jumlah_pemeriksaan += 1

        print(f"Pemeriksaan ke-{jumlah_pemeriksaan}: "
              f"index {i}, nilai {data[i]}")

        if data[i] == target:
            return i, jumlah_pemeriksaan

    return -1, jumlah_pemeriksaan
```

```python
print("Elemen pertama")

target_pertama = data[0]

index, jumlah = linear_search_trace(data, target_pertama)

print("\nTarget:", target_pertama)
print("Index:", index)
print("Jumlah pemeriksaan:", jumlah)
```

```python
print("Elemen terakhir")

target_terakhir = data[-1]

index, jumlah = linear_search_trace(data, target_terakhir)

print("\nTarget:", target_terakhir)
print("Index:", index)
print("Jumlah pemeriksaan:", jumlah)
```

```python
print("Target tidak ditemukan")

target_tidak_ada = 100

index, jumlah = linear_search_trace(data, target_tidak_ada)

print("\nTarget:", target_tidak_ada)
print("Index:", index)
print("Jumlah pemeriksaan:", jumlah)
```

```python
# Data teks
data_nama = ["Bambang", "Ismail", "Andi", "Novi", "Ajix", "Sukmanto"]

def linear_search_text(data, target):
    jumlah_pemeriksaan = 0

    for i in range(len(data)):
        jumlah_pemeriksaan += 1

        print(f"Pemeriksaan ke-{jumlah_pemeriksaan}: {data[i]}")

        if data[i].lower() == target.lower():
            return i, jumlah_pemeriksaan

    return -1, jumlah_pemeriksaan


# Mencari "andi"
index, jumlah = linear_search_text(data_nama, "andi")

print("\nHasil pencarian:")
print("Index:", index)
print("Jumlah pemeriksaan:", jumlah)
```

```python
# Praktik Mandiri 2.3 - Binary Search
# Membuat list minimal 20 angka
data = [
    43, 11, 71, 24, 93,
    39, 87, 5, 26, 79,
    18, 12, 26, 81, 73,
    85, 7, 19, 94, 37
]

# Mengurutkan data dengan sorted()
data_urut = sorted(data)

print("Data awal   :", data)
print("Data terurut:", data_urut)

def binary_search_trace(data, target):
    left = 0
    right = len(data) - 1
    pemeriksaan = 0
    trace = []

    while left <= right:
        mid = (left + right) // 2
        pemeriksaan += 1

        # Simpan nilai untuk tabel trace
        trace.append([pemeriksaan, left, right, mid, data[mid]])

        print(
            f"Pemeriksaan {pemeriksaan}: "
            f"left={left}, right={right}, "
            f"mid={mid}, data[mid]={data[mid]}"
        )

        if data[mid] == target:
            return mid, pemeriksaan, trace

        elif data[mid] < target:
            # Target berada di sebelah kanan
            left = mid + 1

        else:
            # Target berada di sebelah kiri
            right = mid - 1

    return -1, pemeriksaan, trace
```

```python
target_tengah = 87

print("Target di Sekitar Tengah")

indeks, jumlah, trace = binary_search_trace(
    data_urut,
    target_tengah
)

print("\nHasil:")
print("Target          :", target_tengah)
print("Indeks ditemukan:", indeks)
print("Jumlah pemeriksaan:", jumlah)
```
