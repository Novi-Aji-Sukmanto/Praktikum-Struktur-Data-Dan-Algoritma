```python
#3.23. Praktik Mandiri 3.2 — Operasi Insert dan Search
# Class Node
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head

        while current.next is not None:
            current = current.next

        current.next = new_node

    def insert_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def insert_after(self, target, data):
        current = self.head

        while current is not None:
            if current.data == target:
                new_node = Node(data)
                new_node.next = current.next
                current.next = new_node
                return True

            current = current.next

        return False

    def search(self, target):
        current = self.head
        position = 0

        while current is not None:
            if current.data == target:
                return position

            current = current.next
            position += 1

        return -1

    def display(self):
        current = self.head
        hasil = []

        while current is not None:
            hasil.append(str(current.data))
            current = current.next

        print(" -> ".join(hasil) + " -> None")

linked_list = SinglyLinkedList()

linked_list.append(15)
linked_list.append(25)
linked_list.append(35)
linked_list.append(45)

linked_list.insert_beginning(5)

linked_list.insert_after(25, 30)

print("Isi Linked List:")
linked_list.display()
```

```python
# Mencari nilai 30
hasil_30 = linked_list.search(30)

# Mencari nilai 99
hasil_99 = linked_list.search(99)

print("Hasil pencarian:")
print("Nilai 30 ditemukan pada posisi:", hasil_30)
print("Nilai 99 ditemukan pada posisi:", hasil_99)
```

```python
# PRAKTIK MANDIRI 3.3
#DELETE DAN EDGE CASE

# Class Node
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head

        while current.next is not None:
            current = current.next

        current.next = new_node

    def delete(self, target):
        if self.head is None:
            print(f"Nilai {target} tidak bisa dihapus karena list kosong.")
            return False

        if self.head.data == target:
            self.head = self.head.next
            print(f"Nilai {target} berhasil dihapus.")
            return True

        previous = self.head
        current = self.head.next

        while current is not None:
            if current.data == target:
                previous.next = current.next
                print(f"Nilai {target} berhasil dihapus.")
                return True

            previous = current
            current = current.next

        # Jika data tidak ditemukan
        print(f"Nilai {target} tidak ditemukan.")
        return False

    def display(self):
        if self.head is None:
            print("List kosong.")
            return

        current = self.head

        while current is not None:
            print(current.data, end=" -> ")
            current = current.next

        print("None")

    def count(self):
        jumlah = 0
        current = self.head

        while current is not None:
            jumlah += 1
            current = current.next

        return jumlah

    def display_status(self):
        print("Isi list    :", end=" ")
        self.display()
        print("Jumlah node :", self.count())
        print("-" * 40)
```

```python
print("HAPUS PADA LIST KOSONG")

list_kosong = SinglyLinkedList()

print("Kondisi awal:")
list_kosong.display_status()

print("Tes hapus nilai 10:")
list_kosong.delete(10)

print("Kondisi setelah di hapus:")
list_kosong.display_status()
```

```python
print("HAPUS HEAD, TENGAH, TERAKHIR")

linked_list = SinglyLinkedList()

data = [10, 20, 30, 40, 50]

for nilai in data:
    linked_list.append(nilai)

print("Kondisi awal:")
linked_list.display_status()

print("1. Menghapus HEAD (10):")
linked_list.delete(10)
linked_list.display_status()

print("2. Menghapus node tengah (30):")
linked_list.delete(30)
linked_list.display_status()

print("3. Menghapus node terakhir (50):")
linked_list.delete(50)
linked_list.display_status()
```
